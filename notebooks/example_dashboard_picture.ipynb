# Generate a picture with text in the center

```python
text_example = "Could not find picture"
now = datetime.now()
```

```python
width = 960
height = 680
my_dpi = 96
ratio = height / width
```

```python
font = dict(
    fontweight="bold",
    fontfamily="Fira Sans",
)
```

```python
fig, ax = plt.subplots(
    1,1, figsize=(width/my_dpi, height/my_dpi), dpi=my_dpi
)

ax.text(
    .5,
    .55,
    text_example,
    verticalalignment='center',
    horizontalalignment='center',
    fontsize=22,
    **font,
)
ax.text(
    .5,
    .45,
    now.strftime("%Y-%m-%d %H:%M"),
    verticalalignment='center',
    horizontalalignment='center',
    fontsize=15,
    **font,
)

ax.axis('off')

# NOTE: or use ax.set_aspect(ratio)

pass
```

## Save to Pillow image

```python
buf = io.BytesIO()
fig.savefig(buf, dpi=my_dpi)
buf.seek(0)
img = Image.open(buf)
```
